```python
import langchain 
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import FewShotPromptTemplate
import pandas as pd
import streamlit as st
```

```python
from langchain_ollama import OllamaLLM
```

```python
df =pd.read_csv("dataset/train.csv")
```

```python
dict = df[['Question','Answer']].to_dict('index')
```

```python
QnA = [v for i , v in dict.items()]
```

```python
counter = 0
ind = 0
for i in QnA:
    
    if '{' in i['Question']:
        counter+=1
        print(ind)
    if '{' in i['Answer']:
        counter+=1
        print(ind)
    ind+=1
```

```python
QnA[11458]
```

```python
example_prompt = PromptTemplate.from_template("Question: {Question}\nAnswer: {Answer}")

prompt_template = FewShotPromptTemplate(
    examples=QnA,
    example_prompt=example_prompt,
    suffix="Question: {input}",
    input_variables=["input"],
    example_separator="\n\n"  # Optional: adds space between examples
)

```

```python
llm = OllamaLLM(model="gemma3:1b")

chain = prompt_template | llm

chain.invoke({"input":"Who is at risk for Lymphocytic Choriomeningitis (LCM)?"})
```

```python
print("Answer: Lymphocytic Choriomeningitis (LCM) is a rare, rapidly progressive, and often fatal condition in infants and young children. It’s characterized by inflammation of the choroid plexus (a part of the brain that produces cerebrospinal fluid) and the meninges (membranes surrounding the brain and spinal cord). Here’s a breakdown of the risk factors:\n\n*   **Congenital Lymphocytic Choriomeningitis:** This is the primary risk. LCM is a genetic disorder caused by a mutation in the *HLA-B27* gene. This gene is linked to a higher risk of developing the condition.\n*   **Family History:** Having a family member with LCM significantly increases the risk.\n*   **Early Childhood Exposure to Certain Viruses:** Specific viruses, most notably, *West Nile virus* and *Epstein-Barr virus* (EBV), have been implicated as potential triggers or accelerators in some cases.\n*   **Certain Genetic Syndromes:** Certain genetic syndromes, such as the *Sotos Syndrome* and *Mitochondrial Disease* syndromes, increase the risk.\n*   **Exposure to Certain Medications:** Some medications may be linked, though this is an area of ongoing research.\n\n**Important Note:** While the risk is higher with a specific genetic marker (HLA-B27), it's crucial to remember that LCM is a complex condition with a variable etiology.")
```
